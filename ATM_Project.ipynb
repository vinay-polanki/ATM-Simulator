```python
# Credit Amount
def Credit_amt(Balance,transaction_log):
    Credit_Amount=float(input("Enter the amount to credit: "))
    print(f"Amount Credited to account.$ {Credit_Amount}")
    Balance+=Credit_Amount
    transaction_log.append(
                          { "Type": "Credit",
                            "Amount": Credit_Amount,
                            "Date": datetime.now().strftime('%d-%B-%Y %I:%M:%S %p'),
                            "Balance":Balance
                            })
    print(f"Available Balance: {Balance}")
    Credit_Amount=0.0
    return Balance, Credit_Amount
```

```python
                 # Mini Statement
from datetime import datetime

def MINI_STATEMENT(UserName, Debit_Amount, Credit_Amount, Balance, transaction_log):
    print("\n========== MINI STATEMENT ==========")
    print(f"Date           : {datetime.now().strftime('%d-%B-%Y %I:%M:%S %p')}")
    print(f"Hello          : {UserName}")
    print(f"Amount Debited : ${Debit_Amount}")
    print(f"Amount Credited: ${Credit_Amount}")
    print(f"Balance        : ${Balance}")
    print("====================================\n")

    if transaction_log:
        print("========== TRANSACTIONS ==========")
        print("Type    | Amount   | Date                      | Balance")
        print("--------|----------|---------------------------|---------")
        for a in transaction_log:
            print(f"{a['Type']:<7} | ${a['Amount']:<7.2f} | {a['Date']} | ${a['Balance']:.2f}")
        print("\n===== Thank You! Visit again =======")
        print("====================================\n")
    else:
        print("No Records Found.")
```

```python
# Saving to txt file
def stmt(UserName, Debit_Amount, Credit_Amount, Balance, transaction_log):
        statement = f"""========== MINI STATEMENT ==========
        Date           : {datetime.now().strftime('%d-%B-%Y %I:%M:%S %p')}
        Hello          : {UserName}
        Amount Debited : ${Debit_Amount:.2f}
        Amount Credited: ${Credit_Amount:.2f}
        Balance        : ${Balance:.2f}
        ====================================
        
        ========== TRANSACTIONS ==========
           Type   |    Amount   |           Date            | Balance
        ----------|-------------|---------------------------|---------
        
        """
        
        if transaction_log:
            for a in transaction_log:
                statement += f"{a['Type']:<7} | ${a['Amount']:<7.2f} | {a['Date']} | ${a['Balance']:.2f}\n"
        else:
            statement += "No Records Found.\n"
        
        statement += """
        ===== Thank You! Visit again =====
        ==================================
        """
        
        # Save to file
        filename = f"MINI_Statement_{UserName}_{datetime.now().strftime('%d-%b-%Y_%I-%M-%S_%p')}.txt"
        with open(filename, 'w') as file:
            file.write(statement)
            print(f"Statement saved as {filename}")
```
